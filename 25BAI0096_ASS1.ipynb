```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("1000 Sales Records.csv")

print("HEAD")
print(df.head(10))
print("----------------------------------------------------------")

print("TAIL")
print(df.tail(15))
print("----------------------------------------------------------")

print("INFO")
df.info()
print("----------------------------------------------------------")

print("SHAPE")
print(df.shape)
print("----------------------------------------------------------")

print("COLUMNS")
print(df.columns)
print("----------------------------------------------------------")

print("DTYPES")
print(df.dtypes)
print("----------------------------------------------------------")

print("DESCRIBE")
print(df.describe())
print("----------------------------------------------------------")

print(df.duplicated().sum())
print("----------------------------------------------------------")

print("MEAN OF FIRST 50 RECORDS")
print(df.head(50).mean(numeric_only=True))
print("----------------------------------------------------------")

print("STANDARD DEVIATION")
print(df.std(numeric_only=True))
print("----------------------------------------------------------")

print("MISSING VALUES")
print(df.isnull().sum())
print("----------------------------------------------------------")

print("UNIQUE VALUES")
print(df.nunique())
print("----------------------------------------------------------")

print("GROUP BY UNITS SOLD")
print(df.groupby("Units Sold").size())
print("----------------------------------------------------------")

print("SORTED DATA")
df_sorted = df.sort_values(
    by=["Country", "Total Revenue"],
    ascending=[True, False]
)
print(df_sorted.head(10))

print("----------------------------------------------------------")
print("USING ILOC")
print(df.iloc[0])
print("----------------------------------------------------------")

df['Total Profit'].hist()
plt.title("Distribution of Total Profit")
plt.xlabel("Total Profit")
plt.ylabel("Frequency")
plt.show()

plt.scatter(df["Units Sold"], df["Total Profit"])
plt.xlabel("Units Sold")
plt.ylabel("Total Profit")
plt.show()
```
