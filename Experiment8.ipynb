```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the uploaded dataset
df = pd.read_csv('airline-passengers.csv')
```

```python
print(f"Shape of the dataset: {df.shape}")
```

```python
display(df.head())
```

```python
df['month'] = pd.to_datetime(df['month'])
df = df.set_index('month')
print("DataFrame index set to 'month' column.")
display(df.head())
```

```python
display(df.sample(5))
```

```python
plt.figure(figsize=(16, 8))
sns.lineplot(data=df, x=df.index, y='total_passengers')
plt.title('Monthly Airline Passenger Counts Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.grid(True)
plt.show()
```

```python
df['Cumulative_Passengers'] = df['total_passengers'].cumsum()

plt.figure(figsize=(16, 8))
sns.lineplot(data=df, x=df.index, y='Cumulative_Passengers')
plt.title('Cumulative Airline Passenger Counts Over Time')
plt.xlabel('Date')
plt.ylabel('Cumulative Number of Passengers')
plt.grid(True)
plt.show()
```
